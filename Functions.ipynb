```python
def diagnostico_dataframe(df):
    """
    Muestra información estructural y estadística descriptiva de un DataFrame.
    """
    print("--- ESTRUCTURA ---")
    # Dimensiones del dataframe (filas, columnas)
    print(f"Shape:\n{df.shape}\n")

    # Lista de nombres de columnas
    print(f"Columns:\n{df.columns}\n")

    # Tipos de datos de cada columna
    print(f"Dtypes:\n{df.dtypes}\n")

    print("--- ESTADÍSTICAS ---")
    # Resumen estadístico general
    print(f"Describe:\n{df.describe()}\n")

    # Conteo de valores no nulos
    print(f"Count:\n{df.count()}\n")

    # Mediana de las columnas numéricas
    print(f"Median:\n{df.median(numeric_only=True)}\n")

    # Valores mínimos
    print(f"Min:\n{df.min(numeric_only=True)}\n")

    # Valores máximos
    print(f"Max:\n{df.max(numeric_only=True)}\n")

    # Moda (valor más frecuente)
    print(f"Mode:\n{df.mode().iloc[0]}\n")

    # Desviación estándar
    print(f"Std:\n{df.std(numeric_only=True)}\n")

    # Varianza
    print(f"Var:\n{df.var(numeric_only=True)}\n")

    # Simetría (Skewness)
    print(f"Skew:\n{df.skew(numeric_only=True)}\n")

    # Curtosis
    print(f"Kurt:\n{df.kurt(numeric_only=True)}\n")

    # Nota: 'mad' (Mean Absolute Deviation) fue eliminado en versiones recientes de pandas.
    # Se puede calcular manualmente si es necesario:
    # print(f"MAD:\n{(df - df.mean(numeric_only=True)).abs().mean(numeric_only=True)}\n")
```
