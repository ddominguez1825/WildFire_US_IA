````python

import sqlite3
import pandas as pd
import os

# 1. Montar Google Drive
#drive.mount('/content/drive')

# 2. Configurar la nueva ruta del archivo en Drive
# Nota: Generalmente las rutas en Drive empiezan con /content/drive/MyDrive/...
# Usaré la ruta exacta que proporcionaste:
db_path = r'C:\Master IMF\M11_TrabajoFinDeMaster\ProyectoTFM\WildFireCode\DataBase\FPA_FOD_20170508.sqlite'

if not os.path.exists(db_path):
    print(f"Error: El archivo '{db_path}' no existe. Verifica si la carpeta 'WildFires' está en la raíz o dentro de 'MyDrive'.")
else:
    try:
        conn = sqlite3.connect(db_path)

        # DIAGNÓSTICO: Verificar contenido
        cursor = conn.cursor()
        cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
        tables = cursor.fetchall()
        print(f"Tablas encontradas: {tables}")

        if not tables:
            print("La base de datos está vacía.")
        else:
            # Cargar datos y ejecutar la función de diagnóstico previa
            query = """SELECT * FROM Fires LIMIT 1000;""" # Limitamos para evitar saturar memoria inicialmente
            df = pd.read_sql_query(query, conn)
            print("¡Datos cargados con éxito desde Google Drive!")

            # Llamada a la función definida anteriormente

            print("--- ESTRUCTURA ---")
            # Dimensiones del dataframe (filas, columnas)
            print(f"Shape:\n{df.shape}\n")

            # Lista de nombres de columnas
            print(f"Columns:\n{df.columns}\n")

            # Tipos de datos de cada columna
            print(f"Dtypes:\n{df.dtypes}\n")

            print("--- ESTADÍSTICAS ---")
            # Resumen estadístico general
            print(f"Describe:\n{df.describe()}\n")

            # Conteo de valores no nulos
            print(f"Count:\n{df.count()}\n")

            # Mediana de las columnas numéricas
            print(f"Median:\n{df.median(numeric_only=True)}\n")

            # Valores mínimos
            print(f"Min:\n{df.min(numeric_only=True)}\n")

            # Valores máximos
            print(f"Max:\n{df.max(numeric_only=True)}\n")

            # Moda (valor más frecuente)
            print(f"Mode:\n{df.mode().iloc[0]}\n")

            # Desviación estándar
            print(f"Std:\n{df.std(numeric_only=True)}\n")

            # Varianza
            print(f"Var:\n{df.var(numeric_only=True)}\n")

            # Simetría (Skewness)
            print(f"Skew:\n{df.skew(numeric_only=True)}\n")

            # Curtosis
            print(f"Kurt:\n{df.kurt(numeric_only=True)}\n")

    except Exception as e:
        print(f"Error al conectar o consultar la DB: {e}")
    finally:
        if 'conn' in locals():
            conn.close()
````

````python
def diagnostico_dataframe(df):
    """
    Muestra información estructural y estadística descriptiva de un DataFrame.
    """
    print("--- ESTRUCTURA ---")
    # Dimensiones del dataframe (filas, columnas)
    print(f"Shape:\n{df.shape}\n")

    # Lista de nombres de columnas
    print(f"Columns:\n{df.columns}\n")

    # Tipos de datos de cada columna
    print(f"Dtypes:\n{df.dtypes}\n")

    print("--- ESTADÍSTICAS ---")
    # Resumen estadístico general
    print(f"Describe:\n{df.describe()}\n")

    # Conteo de valores no nulos
    print(f"Count:\n{df.count()}\n")

    # Mediana de las columnas numéricas
    print(f"Median:\n{df.median(numeric_only=True)}\n")

    # Valores mínimos
    print(f"Min:\n{df.min(numeric_only=True)}\n")

    # Valores máximos
    print(f"Max:\n{df.max(numeric_only=True)}\n")

    # Moda (valor más frecuente)
    print(f"Mode:\n{df.mode().iloc[0]}\n")

    # Desviación estándar
    print(f"Std:\n{df.std(numeric_only=True)}\n")

    # Varianza
    print(f"Var:\n{df.var(numeric_only=True)}\n")

    # Simetría (Skewness)
    print(f"Skew:\n{df.skew(numeric_only=True)}\n")

    # Curtosis
    print(f"Kurt:\n{df.kurt(numeric_only=True)}\n")

    # Nota: 'mad' (Mean Absolute Deviation) fue eliminado en versiones recientes de pandas.
    # Se puede calcular manualmente si es necesario:
    # print(f"MAD:\n{(df - df.mean(numeric_only=True)).abs().mean(numeric_only=True)}\n")
````

### Cómo usar la función
Una vez que logres cargar el DataFrame (corrigiendo el archivo `.sqlite`), puedes llamarla así:
```python
diagnostico_dataframe(df)
```
